# Download and Inspect Pretrained Phi-3.5-mini Model


```python
from transformers import AutoModelForCausalLM

model_id = 'microsoft/Phi-3.5-mini-instruct'

model = AutoModelForCausalLM.from_pretrained(model_id)
sd_hf = model.state_dict()

for k, v in sd_hf.items():
    print(k, v.shape)
```

## Initialize Custom PhiCausalLM and Load Pretrained Weights
Load the tokenizer and instantiate `PhiCausalLM`, transferring weights from the downloaded Hugging Face model.


```python
from transformers import AutoTokenizer
from modeling_phi import PhiCausalLM, PhiLMConfig

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Instantiate PhiCausalLM with PhiLMConfig and transfer weights
config = PhiLMConfig()
phi_model = PhiCausalLM(config)
phi_model.load_state_dict(model.state_dict(), strict=False)
phi_model.eval()

print("PhiCausalLM initialized and pretrained weights successfully loaded.")
```

## Test Text Generation Using `phi_model.generate`
Generate text directly using the `generate` method on `PhiCausalLM`. It takes a prompt string and `max_tokens`, and returns the detokenized answer string directly.


```python
# Test text generation using phi_model.generate directly
prompt = "The capital of France is"
max_tokens = 10

answer = phi_model.generate(prompt, max_tokens=max_tokens, tokenizer=tokenizer)

print(f"Prompt: {prompt}")
print(f"Answer: {answer}")
```

```python
# Test with another prompt
prompt2 = "Write a haiku about coding:"
answer2 = phi_model.generate(prompt2, max_tokens=20, tokenizer=tokenizer)

print(f"Prompt: {prompt2}")
print(f"Answer:\n{answer2}")
```
